```python
import json

# --- 1. 직렬화 (딕셔너리 → 문자열) ---
alert = {"rule": 5712, "level": 10, "ip": "1.1.1.1"}
s = json.dumps(alert, ensure_ascii=False)
print(s)            # {"rule": 5712, "level": 10, "ip": "1.1.1.1"}
print(type(s))      # <class 'str'>

# --- 2. 역직렬화 (문자열 → 딕셔너리) ---
d = json.loads(s)
print(d["level"], type(d))   # 10 <class 'dict'>

# --- 3. 한글 테스트 (깨지는 것 → 안 깨지는 것 순서로 비교) ---
alert_kr = {"경보": "로그인 실패", "레벨": 10}
print(json.dumps(alert_kr))                      # ensure_ascii 기본값(True) → 한글이 \uXXXX 로 깨짐
print(json.dumps(alert_kr, ensure_ascii=False))  # 한글 그대로

# --- 4. 보기 좋은 출력 ---
print(json.dumps(alert, ensure_ascii=False, indent=2))
```

```python
def severity(level):
    if level >= 10: return "High"
    if level >= 7:  return "Medium"
    return "Low"

print(severity(10))  # High
print(severity(8))   # Medium
print(severity(3))   # Low
```

```python
def greet(name, msg="안녕"):
    return f"{name}, {msg}!"

print(greet("김보안"))          # 김보안, 안녕!
print(greet("이해킹", "반가워")) # 이해킹, 반가워!
```

```python
def severity(level):
    if level >= 10: return "High"
    if level >= 7:  return "Medium"
    return "Low"

def classify(alert):
    sev = severity(alert["level"])
    return {"ip": alert["ip"], "severity": sev}

alerts = [
    {"level": 12, "ip": "1.1.1.1"},
    {"level": 9,  "ip": "2.2.2.2"},
    {"level": 5,  "ip": "3.3.3.3"},
]
for a in alerts:
    print(classify(a))
```
